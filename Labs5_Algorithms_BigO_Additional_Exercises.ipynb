```python
# [CELL 1] STUDENT INFORMATION
LAST_NAME = "Jandusay"
STUDENT_ID = "TUPM-26-4304"
SEED_DIGIT = 4
ID_SUM = 19
FAVORITE_ARTIST = "Arthur Nery"

NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST.replace(" ", ""))

ARTIST_SUM = sum(
    ord(c.upper()) - ord("A") + 1
    for c in FAVORITE_ARTIST
    if c.isascii() and c.isalpha()
)

print("LAST NAME:", LAST_NAME)
print("STUDENT ID:", STUDENT_ID)
print("SEED DIGIT:", SEED_DIGIT)
print("ID SUM:", ID_SUM)
print("FAVORITE ARTIST:", FAVORITE_ARTIST)
print("NAME LENGTH:", NAME_LENGTH)
print("ARTIST LENGTH:", ARTIST_LENGTH)
print("ARTIST SUM:", ARTIST_SUM)
```

```python
# [CELL 2] Generate and Search Incident Records
record_count = 12 + SEED_DIGIT

incident_records = [
    ID_SUM * 100 + ARTIST_SUM + i * (SEED_DIGIT + 1)
    for i in range(record_count)
]

rotation = SEED_DIGIT % len(incident_records)

rotated_records = (
    incident_records[rotation:]
    + incident_records[:rotation]
)

target_incident = incident_records[
    (ID_SUM + ARTIST_SUM) % len(incident_records)
]


def linear_search(data, target):
    comparisons = 0

    for index, value in enumerate(data):
        comparisons += 1

        if value == target:
            return index, comparisons

    return -1, comparisons


target_index, comparison_count = linear_search(
    rotated_records,
    target_incident
)

print("EMERGENCY DISPATCH RECORD SEARCH")
print("Original Records:", incident_records)
print("Rotation Point:", rotation)
print("Rotated Records:", rotated_records)
print("Target Incident:", target_incident)

if target_index != -1:
    print("Target Position (Index):", target_index)
else:
    print("Target Incident: Not Found")

print("Comparison Count:", comparison_count)
print("Time Complexity: O(n) worst case")
print("Auxiliary Space Complexity: O(1)")
```

```python
# [CELL 3] Generate and Merge Enrollment Records
size_a = 8 + SEED_DIGIT
size_b = 8 + (NAME_LENGTH % 5)

dataset_a = [
    SEED_DIGIT * 1000 + ID_SUM * 10
    + i * (ID_SUM % 7 + 1)
    for i in range(size_a)
]

dataset_b = [
    ARTIST_LENGTH * 1000 + NAME_LENGTH * 10
    + i * (NAME_LENGTH % 7 + 1)
    for i in range(size_b)
]


def merge_sorted_lists(a, b):
    i = 0
    j = 0
    comparisons = 0
    merged = []

    while i < len(a) and j < len(b):
        comparisons += 1

        if a[i] <= b[j]:
            merged.append(a[i])
            i += 1
        else:
            merged.append(b[j])
            j += 1

    merged.extend(a[i:])
    merged.extend(b[j:])

    return merged, comparisons


consolidated, merge_comparisons = merge_sorted_lists(
    dataset_a,
    dataset_b
)

print("UNIVERSITY ENROLLMENT RECORDS")
print("Department A Dataset:", dataset_a)
print("Department B Dataset:", dataset_b)
print("Consolidated Dataset:", consolidated)

print(
    "Records Before Consolidation:",
    len(dataset_a) + len(dataset_b)
)

print("Records After Consolidation:", len(consolidated))
print("Comparison Count:", merge_comparisons)

print(
    "Ordering Verified:",
    consolidated == sorted(consolidated)
)

print(
    "Record Count Verified:",
    len(consolidated) == len(dataset_a) + len(dataset_b)
)

print("Time Complexity: O(n + m)")
print("Auxiliary Space Complexity: O(n + m)")
```

```python
# [CELL 4] Generate the Order Datasets
import random
from time import perf_counter

order_count = 40 + SEED_DIGIT * 5

base_orders = [
    ID_SUM * 1000 + ARTIST_SUM + i * (SEED_DIGIT + 1)
    for i in range(order_count)
]

sorted_orders = sorted(base_orders)

reverse_orders = sorted(
    base_orders,
    reverse=True
)

random_orders = base_orders.copy()

random.Random(SEED_DIGIT + ARTIST_SUM).shuffle(
    random_orders
)

print("E-COMMERCE ORDER PROCESSING")
print("Number of Orders:", order_count)
print("Sorted Orders:", sorted_orders)
print("Reverse-Ordered Orders:", reverse_orders)
print("Random Orders:", random_orders)
```

```python
# [CELL 5] Bubble Sort
def bubble_sort(values):
    arr = values.copy()
    comparisons = 0
    movements = 0

    for end in range(len(arr) - 1, 0, -1):
        swapped = False

        for i in range(end):
            comparisons += 1

            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                movements += 3
                swapped = True

        if not swapped:
            break

    return arr, comparisons, movements
```

```python
# [CELL 6] Selection Sort
def selection_sort(values):
    arr = values.copy()
    comparisons = 0
    movements = 0

    for i in range(len(arr) - 1):
        min_index = i

        for j in range(i + 1, len(arr)):
            comparisons += 1

            if arr[j] < arr[min_index]:
                min_index = j

        if min_index != i:
            arr[i], arr[min_index] = arr[min_index], arr[i]
            movements += 3

    return arr, comparisons, movements
```

```python
# [CELL 7] Insertion Sort
def insertion_sort(values):
    arr = values.copy()
    comparisons = 0
    movements = 0

    for i in range(1, len(arr)):
        key = arr[i]
        j = i - 1

        while j >= 0:
            comparisons += 1

            if arr[j] > key:
                arr[j + 1] = arr[j]
                movements += 1
                j -= 1
            else:
                break

        arr[j + 1] = key
        movements += 1

    return arr, comparisons, movements
```

```python
# [CELL 8] Compare All Sorting Algorithms
algorithms = {
    "Bubble Sort": bubble_sort,
    "Selection Sort": selection_sort,
    "Insertion Sort": insertion_sort
}

conditions = {
    "Already Sorted": sorted_orders,
    "Reverse Ordered": reverse_orders,
    "Random Order": random_orders
}

for condition, data in conditions.items():
    print(f"\n--- {condition} ---")

    for name, algorithm in algorithms.items():
        start = perf_counter()

        result, comparisons, movements = algorithm(data)

        elapsed = perf_counter() - start

        print(f"Algorithm: {name}")
        print("Comparisons:", comparisons)
        print("Movements:", movements)
        print(f"Execution Time: {elapsed:.8f} seconds")
        print("Sorted Correctly:", result == sorted(data))
        print()
```
